## 1 - Questão base

Escreva um pseudo-código que receba três números e determine qual é o maior entre eles, vocês não poderão usar uma função pronta, tipo `max()`.

Crie um novo Jupyter e programe a sua função, teste-a com várias combinações de números, incluindo negativos, positivos, etc. Também se colocarmos um texto no lugar do número a função tem que voltar uma mensagem e erro

```python

# função para analisar os números e achar o maior entre eles, sem a função max()
def maior_numero(numeros):
    maior = numeros[0]
    for numero in numeros:
        if numero > maior:
            maior = numero
    return maior


# chamada da função para achar o maior número
numeros = [3,-2,-40]

maior = maior_numero(numeros)

print(f"O maior número entre {numeros} é {maior}.")
```

## 5 - Análise de construção de regressão linear e teste do modelo

Um modelo de regressão linear foi ajustado para prever o preço de um imóvel (Preço), com base nas seguintes variáveis independentes: Área (m²), Número de Quartos, e Idade do Imóvel (anos).

Abaixo está a saída do modelo de regressão linear:

**Coeficientes:**
- Intercepto: 50000
- Área (m²): 2000
- Número de Quartos: 30000
- Idade do Imóvel (anos): -1000
- R2: 0.85

**1. Equação de regressão linear mais variáveis documentadas**

```python
v0 = 50000 # inceptro
varea = 2000 # quanto R$ acresce por m2
vquartos = 30000 # quanto R$ acresce por quarto
didade = -1000 # R$ depreciação por ano de idade do imóvel
r2 = 0.85


def calcular_preco_imovel(area_m2, num_quartos, idade_anos):

    # Cálculo do preço
    preco = (
        v0
        + varea * area_m2
        + vquartos * num_quartos
        + didade * idade_anos
    )
    return preco

```

**2. Coeficiente de área**

De acordo com a entrada de área, para cada unidade de área a ser comprada, será acrescido R$ 2000 ao preço do imovel

**3. Sinal negativo na idade do imovel**

Para cada ano do imovel, será considerada uma depreciação do preço do mesmo, no caso de R$ 1000

**4. Valor de R2**

O modelo possui um alto poder explicativo no cálculo do valor de imoveis, em 85%.

**5. Previsão de valor de imovel**


```python
# Exemplo de uso
preco_estimado = calcular_preco_imovel(area_m2=100, num_quartos=3, idade_anos=10)
print(f"Preço estimado: R$ {preco_estimado:,.2f}")
```
