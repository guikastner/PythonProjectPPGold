# Geração de Tabuada

Entrada de um número e com um while ele faz a tabuda do numeral inserido


```python
def tabuada(numero):
    contador = 1
    resultados = {}
    for contador in range(1, 11):
        multiplicacao = numero * contador
        chave = "valor " + str(contador)
        resultados[chave] = multiplicacao
        contador += 1
    return resultados


numero = 5 
resultados = tabuada(numero)
print(resultados)
```
