## Adivinhador de números

```python
import random

# cálculo do número
numero_aleatorio = random.randint(1, 100)

def validacao (input, numero_aleatorio, i):
    if input > numero_aleatorio:
        print("Seu chute foi maior que o número secreto. - valor", input, "tentativa: ", i)
        fim = False
        return fim
    elif input < numero_aleatorio:
        print("Seu chute foi menor que o número secreto. - valor", input, "tentativa: ", i)
        fim = False
        return fim
    else:
        print("Parabéns, você acertou o número secreto!")
        fim = True
        return fim
    
def chute():
    fim = False
    i = 1
    while fim == False:
        entrada = int(input("Digite um número entre 1 e 100: "))
        fim = validacao(entrada,numero_aleatorio, i) # chama a função de validação
        i = i + 1
    print("Você tentou", i, "vezes.")

print("Bem-vindo ao jogo de adivinhação!")
print("Estou pensando em um número entre 1 e 100.")
print("Tente adivinhar qual é o número!")

teste = chute()
# Código de adivinhação de números entre 1 e 100


```
