## Leitura de Bibliotecas

```python
import pandas as pd
import numpy as np
```

## Leitura de arquivos

```python
carros_file = "https://raw.githubusercontent.com/EduPekUfpr/PythonProject/refs/heads/main/Dados/Auto_original.csv" #caminho do arquivo

df_carro = pd.read_csv(carros_file)
print(df_carro.head())  # Exibe as primeiras 5 linhas do DataFrame
```

## Exibição dos dados importados

```python
# Análise de dados para entender o que está acontecendo com os carros
# quantidades de linhas e colunas
# Colocando a frase que demontre a quantidade de linhas e colunas no print
print(f"Linhas: {df_carro.shape[0]}, Colunas: {df_carro.shape[1]}")


# Exibe os tipos de dados de cada coluna
print(df_carro.dtypes)  # Exibe os tipos de dados de cada coluna
```

## Ajuste dos nomes de colunas conforme referência fornecida

```python
# Análise dos nomes das colunas da tabela
print(df_carro.columns)  # Exibe os nomes das colunas
```

Ajuste nos nomes e tipos

```python
colunas = ['symboling', 'normalized-losses', 'make', 'fuel-type', 'aspiration', 'num-of-doors', 'body-style', 'drive-wheels', 'engine-location', 'wheel-base', 'length', 'width', 'height', 'curb-weight', 'engine-type', 'num-of-cylinders', 'engine-size', 'fuel-system', 'bore', 'stroke', 'compression-ratio', 'horsepower', 'peak-rpm', 'city-mpg', 'highway-mpg', 'price']


# Fazendo a leitura do df e atribuir o nome das colunas conforme a lista acima
df_carro.columns = colunas
# Exibe os nomes das colunas

df_carro['normalized-losses'] = pd.to_numeric(df_carro['normalized-losses'], errors='coerce')
df_carro['wheel-base'] = pd.to_numeric(df_carro['wheel-base'], errors='coerce')
df_carro['length']= pd.to_numeric(df_carro['length'], errors='coerce')
df_carro['width'] = pd.to_numeric(df_carro['width'], errors='coerce')
df_carro['price'] = pd.to_numeric(df_carro['price'], errors='coerce')
df_carro['curb-weight'] = pd.to_numeric(df_carro['curb-weight'], errors='coerce')
df_carro['engine-size'] = pd.to_numeric(df_carro['engine-size'], errors='coerce')
df_carro['horsepower'] = pd.to_numeric(df_carro['horsepower'], errors='coerce')
df_carro['bore'] = pd.to_numeric(df_carro['bore'], errors='coerce')
df_carro['stroke'] = pd.to_numeric(df_carro['stroke'], errors='coerce')
df_carro['compression-ratio'] = pd.to_numeric(df_carro['compression-ratio'], errors='coerce')
df_carro['horsepower'] = pd.to_numeric(df_carro['horsepower'], errors='coerce')
df_carro['peak-rpm'] = pd.to_numeric(df_carro['peak-rpm'], errors='coerce')
df_carro['city-mpg'] = pd.to_numeric(df_carro['city-mpg'], errors='coerce')
df_carro['highway-mpg'] = pd.to_numeric(df_carro['highway-mpg'], errors='coerce')
print(df_carro.dtypes)  # Exibe os tipos de dados de cada coluna

```

```python
print(df_carro.head())  # Exibe as primeiras 5 linhas do DataFrame
```

## Checagem de dados faltantes

```python
# Exibir qualquer linha com ? em qualquer coluna
print(df_carro[df_carro.isin(['?']).any(axis=1)])  # Exibe qualquer linha com ? em qualquer coluna
```

```python
# Exibir quantidade de NaN em cada coluna ordenando por quantidade de NaN
print(df_carro.isnull().sum().sort_values(ascending=False))  # Exibe quantidade de NaN em cada coluna ordenando por quantidade de NaN
print('- - - ')

```

```python
# Exibir percentual de NaN em cada coluna ordenando por maior percentual de NaN
print(df_carro.isnull().sum().sort_values(ascending=False) / len(df_carro))  # Exibe percentual de NaN em cada coluna ordenando por maior percentual de NaN
print('- - - ')
```

## Limpeza de dados

```python
# exclusão da coluna 'normalized-losses' por ter muito NaN
df_carro.drop(columns=['normalized-losses'], inplace=True)

# Exclusão das linhas que contenham price NaN
df_carro.dropna(subset=['price'], inplace=True)

# contagem de itens por coluna que possuem NaN ordene por crescente e apenas exibir as colunas que possuem NaN
print(df_carro.isnull().sum().sort_values(ascending=False))  # Exibe contagem de itens por coluna que possuem NaN ordene por crescente e apenas exibir as colunas que possuem NaN
print('- - - ')
```

## Pivot Table

```python
# Criar uma nova tabela (PIVOT) onde as linhas serão o fabricante (`make`) as colunas o tipo de carroceria (`body-style`) e os valores são os valores da coluna price médios dos respectivos carros.
df_pivot = df_carro.pivot_table(index='make', columns='body-style', values='price', aggfunc='mean')
print(df_pivot)  # Exibe as primeiras 5 linhas do DataFrame pivot
```
