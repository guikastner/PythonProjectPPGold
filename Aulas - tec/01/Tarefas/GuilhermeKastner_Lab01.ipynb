## Manipulação da primeira string com uma variável de apoio

```python
str01_manipulada = str01
print(str01_manipulada)
len(str01_manipulada)

# tentando alterar o primeiro caracter da string
try:
    str01_manipulada[0] = 'h'
except:
    print('Não é possível alterar uma string')

# manipulando criando uma nova string

str01_manipulada2 = 'h' + str01_manipulada[1:]
print(str01_manipulada2)

# Definindo uma string de exemplo
str01_espacos = '  '+str01_manipulada2+ '    '
print(str01_espacos)
```
