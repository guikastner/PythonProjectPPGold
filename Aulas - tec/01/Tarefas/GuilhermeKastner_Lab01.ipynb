### String

* Em um segundo bloco de código, crie duas strings 'STR01' e 'STR02' e execute as operações de string dos vídeos tutoriais.

```python
str01 = 'Python é incrível'
str02 = 'Teste Str'
```

Testando os parâmetros de string

```python
def exibir_info_string(entrada):
    print(entrada)
    print(len(entrada))

# Exemplo de uso:
exibir_info_string(str01)
print("")
exibir_info_string(str02)
```

Mudando parâmetros

```python
def corrigir_primeiro_caracter(original):
    try:
        novo_caracter = 'x'
        original[0] = novo_caracter
        print(original)
        return original
    except Exception as e:
        return f"Erro ao tentar modificar a string: {e}"

# Exemplo de uso:
corrigir_primeiro_caracter(str01)
corrigir_primeiro_caracter(str02)
```

Criando uma nova string para emular a alteração

```python
#Criando uma nova string com a alteração desejada
def corrigir_primeiro_caracter_nstring(original):
    try:
        novo_caracter = 'x'
        original = novo_caracter + original[1:]
        print(original+ " - ")
        return original
    except Exception as e:
        return f"Erro ao tentar modificar a string: {e}"

# Exemplo de uso:
corrigir_primeiro_caracter_nstring(str01)
corrigir_primeiro_caracter_nstring(str02)
```

Atribuição em cima da mesma variável

```python
# atribuir a alteração de string na mesma variável

s = "Python"
s = 'J' + s[1:]

print(s) # Jython
```

Manipulação de string

```python
#defnindo uma string de exemplo
s = "  Python é incrível  "

#Removendo espaços em branco no início e no fim da string
s_strip = s.strip()
print(f"strip(): '{s_strip}'")
```

Converter as strings em maiúsculo ou minúsculo

```python
def exibir_transformacoes(s_input):
    """
    Recebe uma string, exibe a versão em caixa alta e a versão em caixa baixa.
    """
    s_upper = s_input.upper()
    print(f"upper(): '{s_upper}'")
    
    s_lower = s_input.lower()
    print(f"lower(): '{s_lower}'")

# Exemplo de uso:
exibir_transformacoes(str01)
print("")
exibir_transformacoes(str02)
```

Substituir parte de uma string por outra

```python
def exibir_replace(s_input):
    s_replace = s_input.replace('incrível', 'fantástico')
    print(f"replace(): '{s_replace}'")

# Exemplo de uso:
exibir_replace(str01)
print("")
exibir_replace(str02)
```

Testar início e final de string

```python
def exibir_starts_ends(s_input):
    starts_with = s_input.startswith('Python')
    print(f"startswith('Python'): {starts_with}")
    
    ends_with = s_input.endswith('vel')
    print(f"endswith('vel'): {ends_with}")

# Exemplo de uso:
exibir_starts_ends(str01)
print("")
exibir_starts_ends(str02)
```

Separação e junção de strip

```python
def exibir_split(s_input):
    s_split = s_input.split()
    print(f"split(): {s_split}")

# Exemplo de uso:
exibir_split(str01)
print("")
exibir_split(str02)
```

Encontrar a posição de alteração na lista

```python
def exibir_find(s_input):
    position = s_input.find('é')
    print(position)

exibir_find(str01)
print("")
exibir_find(str02)
```

Contar uma determinada substring

```python
def exibir_count(s_input):
    count = s_input.count('i')
    print(f"count('i'): {count}")

# Exemplo de uso:
exibir_count(str01)
print("")
exibir_count(str02)
```
